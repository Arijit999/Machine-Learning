```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix
from sklearn import tree
import matplotlib.pyplot as plt 
import pandas as pd
```

## Data Preprocessing

```python
data = load_iris()
X = pd.DataFrame(data.data,columns=data.feature_names)
Y = data.target
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.3)
```

## Trainig

```python
model = DecisionTreeClassifier(ccp_alpha=0.01)
model.fit(X_train,Y_train)
```

## Predicting

```python
#0 : Iris Setosa.
#1 : Iris Versicolor.
#2 : Iris Virginica.
Y_pred = model.predict(X_test)
Y_pred
```

## Evaluation

```python
# Acuraccy of model
accuracy_score(Y_test,Y_pred)*100
```

```python
confusion_matrix(Y_test,Y_pred,labels=[0,1,2])
```

```python
imp_features = pd.DataFrame(model.feature_importances_*100,index=data.feature_names,columns=["Percent"])
imp_features
```

## Graph plotting

```python
_ = imp_features.plot(kind="bar")
```

```python
plt.figure(figsize=(20,22))
_ = tree.plot_tree(model,feature_names = data.feature_names,
                   class_names = {0:"Iris Setosa",1:"Iris Versicolor",2:"Iris Virginica"},
                   filled=True,fontsize=20)
```
